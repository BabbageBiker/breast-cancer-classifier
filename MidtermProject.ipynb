```python
import pandas as pd
# benign/malignant target feature = categorical
# means this is a Classification task.
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import f1_score

import numpy as np
```

```python
# loading and cleaning Data.
data = pd.read_csv('breast-cancer.csv')
# encode categorical target(maligant, benign) --> numerical target(1, 0)
data['diagnosis'] = data['diagnosis'].map({'M': 1, 'B': 0})
numericalFeatures = data.select_dtypes(include = ['float64', 'int64']).drop(columns = ['diagnosis']) 

data.fillna(value = 0, inplace = True) # fill any empty data

# trim as needed
if len(data) > 10000:
    data = data.sample(n = 10000, random_state = 42)
    data.to_csv('breast-cancer-sample.csv', index = False)

# 70% / 30% train test split
trainData, testData = train_test_split(data, test_size = 0.3, random_state = 42)

targetFeature = 'diagnosis' # whether tumor is benign or malignant(0, 1)
trainXs = trainData.drop(columns = [targetFeature])
trainYs = trainData[targetFeature]
testXs = testData.drop(columns = [targetFeature])
testYs = testData[targetFeature]

print("Training fts (trainXs):", trainXs.shape)
print("Training target (trainYs):", trainYs.shape)
print("Testing fts (testXs):", testXs.shape)
print("Testing target (testYs):", testYs.shape)

#print(data.info())
```

```python
# creating and fitting a pipeline. (classification model)
pipeline = Pipeline(steps = [('scale', MinMaxScaler()),
    ('classify', LogisticRegression(max_iter = 1000, random_state = 42))
])
pipeline.fit(trainXs, trainYs)
```

```python
# predicting and scoring.
# Logistic Regression for binary classification
# F1-Score for balance of precision and recall
 # minimize false negatives (+false positives)

predictedYs = pipeline.predict(testXs)
f1Score = f1_score(testYs, predictedYs)
print(f"F1-Score = {f1Score:.4f}")
```

The second reason F1 score was appropriate for this task was because the F1 score helped ensure that the model was correctly identifying tumors as maligant, but also ensuring that the model is avoiding identifying benign tumors as malignant.
